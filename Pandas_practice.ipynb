```python
import numpy as np
import pandas as pd
```

```python
#1

list = pd.Series([10, 20, 30, 40, 50])
print(list)
```

```python
#2
marks_list = pd.Series({'English':80, 'Math':90, 'Science':70, 'Hindi':95})
print(marks_list)
```

```python
#3
df = pd.DataFrame([
    {'Name':'English', 'Marks':80},
    {'Name':'Math', 'Marks':90},
    {'Name':'Science', 'Marks':70},
    {'Name':'Hindi', 'Marks':85}
])
print(df)
```

```python
#4

df['Country'] = 'India'
print(df)
```

```python
#5

Stud_name = pd.Series(['Jim', 'Pam', 'Erin', 'Dwight'])
df['Student_name'] = Stud_name
```

```python
print(df)
```

```python
#6.1

df.rename(columns={'Name':'Subjects'}, inplace=True)
print(df)
```

```python
#6.2

df.rename(columns={
    'Marks':'Score',
    'Student_name':'Students'
}, inplace = True)

print(df)
```

```python
#7

employees = pd.DataFrame({
    'Name':['John','Mary','Alice','Bob'],
    'Department':['IT','HR','IT','Finance'],
    'Salary':[50000,60000,55000,70000]
}, index=['E1','E2','E3','E4'])

print(employees)
```

```python
#single rows

print(employees.loc['E1'])
```

```python
#multiple rows

print(employees.loc[['E2','E3']])
```

```python
#conditional rows

print(employees.loc[employees['Salary'] >= 60000])
```

```python
#8

emp = pd.DataFrame({
    'Name':['Jim','Pam','Mary','Michael'],
    'Department':['IT','HR','IT','Finance'],
    'Salary':[40000,60000,55000,80000]
}, index=['1','2','3','4'])

print(emp)
```

```python
#First row

print(emp.iloc[0])
```

```python
#second col

print(emp.iloc[:,1])
```

```python
#First 3 rows and 2 col

print(emp.iloc[:3, :2])
```

```python
#9
#shape and datatype

df = pd.DataFrame({
    'Movie':['Iron Man', 'Avengers', 'Transformers', 'Ant Man'],
    'Release':[2008, 2012, 2007, 2015],
    'Actor':['Robert', 'Chris', 'Shia', 'Paul']
}, index = ['1', '2', '3', '4'])

print(df)
```

```python
print("Shape of dataframe:", df.shape)
```

```python
print("Datatype of dataframe:\n",df.dtypes)
```

```python
#10
sample = pd.DataFrame({
    'a':[10, 20, np.nan],
    'b':[np.nan, 30, 40]
})

print(sample)
```

```python
print(sample.fillna(0))
```
