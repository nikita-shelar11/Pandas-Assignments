# Pandas Complete Tutorial — Beginner to Advanced

## 1. Series and DataFrame

Series is 1-D labeled data; DataFrame is a 2-D table.


```python
import pandas as pd
import numpy as np
s=pd.Series([10,20,30],index=["A","B","C"])
df=pd.DataFrame({"name":["Amit","Neha","Raj"],"age":[22,25,24],"score":[85,92,78]})
print(s); print(df)
```

## 2. Inspection

Use head, tail, shape, columns, dtypes, info and describe to understand data.


```python
print(df.head())
print(df.shape)
print(df.columns.tolist())
print(df.dtypes)
print(df.describe())
```

## 3. loc and iloc

loc uses labels/conditions; iloc uses integer positions.


```python
print(df.loc[df["score"]>80,["name","score"]])
print(df.iloc[0:2,0:2])
```

## 4. Filtering

Boolean filtering is fundamental for analysis.


```python
print(df[df["age"]>=24])
print(df[(df["age"]>=24)&(df["score"]>=80)])
```

## 5. Columns and feature engineering

Create derived columns for analysis.


```python
df["passed"]=df["score"]>=40
df["score_plus_5"]=df["score"]+5
print(df)
```

## 6. Missing values and duplicates

Inspect missing data, impute where appropriate, and remove duplicates.


```python
dirty=pd.DataFrame({"name":["A","B","B","C"],"score":[90,np.nan,np.nan,75]})
print(dirty.isna().sum())
dirty["score"]=dirty["score"].fillna(dirty["score"].mean())
print(dirty.drop_duplicates())
```

## 7. Sorting

sort_values and sort_index provide deterministic ordering.


```python
print(df.sort_values("score",ascending=False))
```

## 8. GroupBy and aggregation

GroupBy follows split → apply → combine.


```python
sales=pd.DataFrame({"city":["Pune","Pune","Mumbai","Mumbai","Pune"],"sales":[100,150,200,120,80]})
print(sales.groupby("city")["sales"].agg(["sum","mean","count"]))
```

## 9. Merge, join and concat

Combine related tables using keys or stack tables.


```python
customers=pd.DataFrame({"id":[1,2,3],"name":["A","B","C"]})
orders=pd.DataFrame({"id":[1,1,2,3],"amount":[100,200,150,300]})
print(customers.merge(orders,on="id",how="left"))
print(pd.concat([customers,customers],ignore_index=True))
```

## 10. Datetime

Datetime features help with trends and time-based analysis.


```python
dates=pd.to_datetime(["2026-01-10","2026-02-15","2026-03-20"])
print(dates.month)
print(dates.day_name())
```

## 11. String operations

The .str accessor provides vectorized text operations.


```python
names=pd.Series(["alice smith","BOB JONES","charlie"])
print(names.str.title())
print(names.str.len())
print(names.str.contains("bob",case=False))
```

## 12. apply, map and lambda

Use vectorized operations where possible; apply/map are useful for custom logic.


```python
scores=pd.Series([45,72,88,33])
print(scores.map(lambda x:"Pass" if x>=40 else "Fail"))
```

## 13. Pivot and melt

Pivot summarizes data; melt converts wide data into long format.


```python
wide=pd.DataFrame({"city":["Pune","Mumbai"],"Jan":[100,200],"Feb":[120,180]})
print(wide.melt(id_vars="city",var_name="month",value_name="sales"))
```

## 14. Rolling/window operations

Rolling and expanding calculations are useful in time-series analysis.


```python
x=pd.Series([10,20,30,40,50])
print(x.rolling(3).mean())
print(x.expanding().mean())
```

## 15. Statistics and correlation

Pandas supports descriptive statistics and correlation analysis.


```python
data=pd.DataFrame({"hours":[2,4,5,7,8],"score":[50,60,65,78,90]})
print(data.describe())
print(data.corr(numeric_only=True))
```

## 16. End-to-end mini EDA

A standard workflow is load → inspect → clean → transform → analyze.


```python
df2=pd.DataFrame({"department":["IT","IT","HR","Sales","Sales"],"salary":[60000,70000,np.nan,50000,65000],"experience":[2,4,3,1,5]})
df2["salary"]=df2["salary"].fillna(df2["salary"].median())
print(df2.groupby("department").agg(avg_salary=("salary","mean"),avg_experience=("experience","mean"),employees=("salary","size")))
```
