```python
import numpy as np
import pandas as pd
```

```python
### NP1. Create a NumPy array and convert it to a Pandas Series.

#> 💡 **Hint:** `np.array([...])` creates the array. Pass it directly to `pd.Series(array, index=[...], name="...")`.

arr = np.array([10, 20, 30, 40, 50])
s = pd.Series(arr, index = ['1','2','3','4','5'], name = 'Numbers')
print(s)
```

```python
### NP2. Convert a 2D NumPy array to a Pandas DataFrame.

#> 💡 **Hint:** `pd.DataFrame(np_2d_array, columns=[...])` — shape `(n, m)` gives `n` rows and `m` columns.

arr_2d = np.array([
    ['Nikita', 90],
    ['Avinash', 80],
    ['Chetan', 85],
    ['Sakshi', 92]
])

df = pd.DataFrame(arr_2d, columns = ['Name', 'Percentage'], index = ['1', '2','3','4'])

print(df)
```

```python
### NP3. Generate a random dataset using NumPy and store it in a DataFrame.

#> 💡 **Hint:** Use `np.random.seed()` for reproducibility. Try `np.random.randn(n, m)` for normal distribution or `np.random.randint(lo, hi, n)` for integers.

dataset = np.random.randint(1, 10, 5)

dataframe = pd.DataFrame(dataset, columns = ['Numbers'])

print(dataframe)
```

```python
### NP4. Perform element-wise operations using NumPy and Pandas.

#> 💡 **Hint:** You can apply NumPy ufuncs directly on a Pandas Series: `np.sqrt(df["col"])`. The result is a Series.

arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([10, 20, 30, 40, 50])
```

```python
#Element wise operations on numpy arrays

print("Addition of array:", arr1 + arr2)
print("Subtraction of array:", arr2 - arr1)
print("Multiplication of array:", arr1 * arr2)
print("Division of array:", arr1/arr1)
print("Power of array:", arr2 ** arr1)
```

```python
#Element wise operations on pandas series

a = pd.Series([1, 2, 3, 4, 5])
b = pd.Series([10, 20, 30, 40, 50])

print(a + b)
print(b - a)
print(a * b)
print(b / a)
print(b ** a)
```

```python
#element wise operations on pandas dataframe

df1 = pd.DataFrame({'A': [1, 2, 3, 4, 5], 'B': [10, 20, 30, 40, 50]})
df2 = pd.DataFrame({'A': [1, 2, 3, 4, 5], 'B': [100, 200, 300, 400, 500]})

print(df1 + df2)
print(df2 - df1)
print(df1 * df2)
print(df2 / df1)
print(df2 ** df1)
```

```python
### NP5. Apply `np.mean` on DataFrame columns.

#> 💡 **Hint:** `np.mean(df["col"])` works the same as `df["col"].mean()`. Apply to the `.values` array for pure NumPy speed: `np.mean(df["col"].values)`.

df = pd.DataFrame({
    "Name" : ["Alice", "Pam", "Jim", "Mandy"],
    "Score" : [90, 85, 95, 75]
})
print("Mean of student scores using pandas:", df["Score"].mean())
print("Mean of student scores using numy:", np.mean(df["Score"].values))
```

```python
### NP6. Replace values using `np.where`.

#> 💡 **Hint:** `np.where(condition, value_if_true, value_if_false)` — assign the result back to a column: `df["col"] = np.where(df["col"] > 50, "High", "Low")`.

df = pd.DataFrame({
    "Name" : ["Alice", "Pam", "Jim", "Mandy"],
    "Score" : [30, 95, 35, 75]
})

result = np.where(df["Score"] > 50, "High", "Low")
print(result)
```
